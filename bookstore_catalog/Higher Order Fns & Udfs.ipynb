```sql
-- filter function
USE ecommerce_catalog.bookstore;

SELECT *,
  filter(books, i -> i.quantity >= 2) As books
FROM Orders;
```

```sql
-- above query we are getting empty arrays in books column. In this we are removing that
SELECT * FROM 
(
SELECT Order_Id,
    quantity,
  filter(books, i -> i.quantity >= 2) As multiple_copies
  FROM Orders
) AS T
WHERE size(multiple_copies) >0
```

```sql
-- Transform function

SELECT order_id,
  customer_id,
  books,
  transform(books, i -> i.subtotal * 0.8) as discount
FROM orders
```
