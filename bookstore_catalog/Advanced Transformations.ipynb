```sql
use ecommerce_catalog.bookstore;

select * from orders;
```

```sql
DESCRIBE Orders
```

```sql
SELECT order_id,books.bookId
FROM orders
```

```sql
Create table explode_orders AS 
SELECT *, explode(books)as books_split FROM Orders
```

```sql
SELECT  order_id,
  books_split.bookId,
  books_split.quantity,
  books_split.subtotal
FROM explode_orders
```

```sql
-- get unique bookId for customer
select customer_id,
 collect_set(books.bookId)
 from orders
 group by customer_id
```

```sql
-- get unique bookId for customer even though if the numbers exists in multiple sets
-- use array distinct and flatten methods
select customer_id,
       array_distinct(flatten(collect_set(books.bookId))) as distinct_books
 from orders
 group by customer_id
```

```sql
-- join clause for each order using explode and books table

select * 
from (
  select *, explode(books) as book from orders
)as od
Inner join books bk
  on bk.book_Id = od.book.bookId
```
