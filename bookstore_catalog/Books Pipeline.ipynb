```sql
--bronze layer
create or refresh streaming live table books_bronze
as select * from cloud_files("dbfs:/Volumes/ecommerce_catalog/default/bookstreams/pipelineDemo/cdc/")
```

```sql
-- silver layer
-- create or refresh streaming live table books_silver;
Apply changes into live.books_silver
from stream (Live.books_bronze)
keys (book_Id)
apply as delete when row_status = "delete"
sequence by row_time
columns * except (row_status,row_time)
```

```sql
-- gold layer
create live table author_counts_sale
as select author, count(*) as count from Live.books_silver group by Author
```

```sql
-- DLT View
create live view bookssales
as
select *
from Live.author_counts_sale
```
